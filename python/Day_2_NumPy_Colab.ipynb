# Vectorized Operations

Vectorization means applying an operation to **many values at once** instead of manually looping through them.

---

## 20. Convert centimeters to millimeters

The Iris measurements are in centimeters.

Since:


## 21. Try other vectorized functions

### What are we doing?

NumPy can apply mathematical functions across the whole array.

We'll use only the first flower so the output stays easy to read.

````python
flower = X[0]

print("Original: ", flower)
print("Add 1:    ", np.add(flower, 1))
print("Subtract:", np.subtract(flower, 1))
print("Multiply:", np.multiply(flower, 2))
print("Divide:  ", np.divide(flower, 2))
print("Square:  ", np.power(flower, 2))
print("Sqrt:    ", np.sqrt(flower))
````

### What this means

Each NumPy function is applied element by element across the first flower's four measurements. This is vectorized computation: one operation acts on the whole array.

# Broadcasting

Broadcasting lets NumPy use a smaller array across a larger compatible array.

A practical example is **mean centering**.

---

## 28. Calculate the average of each column

````python
feature_means = X.mean(axis=0)

feature_means
````

### What this means

`feature_means` contains four numbers: the average sepal length, sepal width, petal length, and petal width.

## 29. Subtract those averages from every flower

### What are we doing?

`X` has 150 rows, but `feature_means` has only 4 values.

NumPy automatically applies those four values to every row.

That is broadcasting.

````python
centered = X - feature_means

centered[:5]
````

### What this means

NumPy broadcasts the four feature means across all 150 rows and subtracts the correct mean from each column. This is called mean centering.

If we calculate the column averages again, they should now be approximately zero.

````python
centered.mean(axis=0)
````

### What this means

The four results should be extremely close to zero. That confirms each column was centered around its mean.

# Matrix Multiplication

Matrix multiplication combines rows and columns.

In NumPy:

```python
A @ B
```

or:

```python
np.matmul(A, B)
```

---

## 33. Create a tiny weight matrix

### What are we doing?

Each Iris flower has **4 features**.

We'll create weights that transform those 4 values into **2 new values**.

````python
weights = np.array([
    [0.2, 0.5],
    [0.4, 0.1],
    [0.6, 0.3],
    [0.8, 0.7]
])

weights
````

### What this means

This creates a `4 × 2` weight matrix. Each flower has 4 input features, and these weights will transform them into 2 output values.

## 34. Multiply 5 flowers by the weights

### What are we doing?

Our shapes are:

```text
5 × 4
@
4 × 2
```

The result will be:

```text
5 × 2
```

````python
X[:5] @ weights
````

### What this means

Five flowers with 4 features each are multiplied by a `4 × 2` matrix. The output therefore has shape `5 × 2`: two new values for each flower.

# Copy vs View

NumPy slices can sometimes share memory with the original array.

Let's see it with a tiny example.

---

## 40. Create a view

````python
demo = np.array([10, 20, 30, 40])

view = demo[1:3]

view
````

### What this means

`view` contains part of `demo`, but it may share the same underlying memory rather than owning separate data.

## 41. Change the view

### What are we checking?

If the view shares memory with `demo`, changing the view can also change `demo`.

````python
view[0] = 999

demo
````

### What this means

The original array changed because the slice was a view. Both arrays were referring to the same underlying data.

## 42. Create an independent copy

### What are we doing?

`.copy()` makes separate data.

Changing the copy will not change the original.

````python
demo = np.array([10, 20, 30, 40])

independent = demo.copy()
independent[0] = 999

print("Original:", demo)
print("Copy:", independent)
````

### What this means

The copied array owns separate data. Changing `independent` does not change `demo`.
