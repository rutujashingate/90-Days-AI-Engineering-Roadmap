# 29. String operations

### Before you run the code

Text columns use the `.str` accessor for vectorized string operations.

````python
df["pickup_borough"].str.lower().head()
````

### What this output means

`str.lower()` converts each text value to lowercase.

# 30. Search text with str.contains()

### Before you run the code

`str.contains()` checks whether text contains a pattern.

````python
df["pickup_zone"].str.contains("Airport", case=False, na=False).head()
````

### What this output means

You get `True` or `False` per row. `na=False` treats missing text as `False`.

# 31. Parse date/time columns

### Before you run the code

CSV date columns may load as text. Use `pd.to_datetime()` to convert them into real datetime values.

````python
df["pickup"] = pd.to_datetime(df["pickup"])
df["dropoff"] = pd.to_datetime(df["dropoff"])
df[["pickup", "dropoff"]].dtypes
````

### What this output means

The columns are now datetime types, so pandas can extract hours/dates and calculate time differences.

# 32. Extract pickup hour

### Before you run the code

The `.dt` accessor exposes date/time components.

````python
df["pickup_hour"] = df["pickup"].dt.hour
df[["pickup", "pickup_hour"]].head()
````

### What this output means

`pickup_hour` is a new numerical feature extracted from the timestamp.

# 33. Calculate trip duration

### Before you run the code

Datetime columns can be subtracted from each other.

````python
df["trip_duration"] = df["dropoff"] - df["pickup"]
df[["pickup", "dropoff", "trip_duration"]].head()
````

### What this output means

`trip_duration` stores the time elapsed between pickup and drop-off.

# 34. Group data

### Before you run the code

`groupby()` splits rows into groups based on a category.

````python
df.groupby("pickup_borough")["fare"].mean()
````

### What this output means

pandas groups trips by borough, selects `fare`, and computes one mean fare per borough.

# 35. Multiple aggregations

### Before you run the code

`agg()` calculates several summary statistics at once.

````python
df.groupby("pickup_borough")["fare"].agg(["count", "mean", "min", "max"])
````

### What this output means

Each borough now has trip count, average fare, lowest fare, and highest fare.

# 36. apply() awareness

### Before you run the code

`apply()` runs a custom function over values. Built-in vectorized operations are often preferable when they exist.

````python
df["fare_category"] = df["fare"].apply(
    lambda fare: "High" if fare >= 20 else "Regular"
)
df[["fare", "fare_category"]].head()
````

### What this output means

Each fare was mapped to `"High"` or `"Regular"` using a custom function.

# 37. map() awareness

### Before you run the code

`map()` is useful for converting known values to other values.

````python
payment_map = {
    "credit card": "Card",
    "cash": "Cash"
}
df["payment_short"] = df["payment"].map(payment_map)
df[["payment", "payment_short"]].head(10)
````

### What this output means

Known payment labels were converted using the dictionary. Values not in the dictionary become `NaN`.

# 38. Create a second DataFrame for merging

### Before you run the code

A merge combines separate tables using a shared key.

````python
borough_info = pd.DataFrame({
    "pickup_borough": ["Manhattan", "Queens", "Brooklyn", "Bronx"],
    "region_code": ["M", "Q", "BK", "BX"]
})
borough_info
````

### What this output means

This second DataFrame shares the `pickup_borough` column with the taxi data.

# 39. Merge / left join

### Before you run the code

Use `merge()` to combine related tables.

````python
merged = df.merge(
    borough_info,
    on="pickup_borough",
    how="left"
)
merged[["pickup_borough", "region_code"]].head()
````

### What this output means

A left join keeps all taxi rows and adds `region_code` where a borough match exists.

Common joins:
- `inner` → matching rows only
- `left` → all left-table rows
- `right` → all right-table rows
- `outer` → rows from both tables

# 41. Build a cleaned DataFrame

### Before you run the code

Let's make a cleaned copy before saving.

## 41.1 Make a copy of the DataFrame

### Before you run the code

Before cleaning the data, we'll create a copy of the original DataFrame.

This is useful because it lets us experiment with cleaning steps without changing the original `df`.

```python
df.copy()
```

creates a separate DataFrame with the same data.

````python
cleaned_df = df.copy()
````

### What this output means

There may not be any visible output from this cell.

`cleaned_df` now contains the same data as `df`, but we can modify `cleaned_df` while keeping the original `df` unchanged.

## 41.2 Remove duplicate rows

### Before you run the code

A **duplicate** is a row that appears more than once.

Duplicate records can affect counts, averages, and later model training.

`drop_duplicates()` removes repeated rows while keeping one copy.

````python
cleaned_df = cleaned_df.drop_duplicates()

cleaned_df.shape
````

### What this output means

`cleaned_df.shape` shows the number of rows and columns after duplicates were removed.

If the row count decreased, duplicate rows were removed.

If the row count stayed the same, there were no exact duplicate rows in the dataset.

## 41.3 Remove rows missing important values

### Before you run the code

Not every missing value means a row must be deleted.

For this exercise, we'll only remove rows where one of these important columns is missing:

- `fare`
- `distance`
- `pickup_borough`

The `subset=` argument tells pandas which columns to check.

````python
cleaned_df = cleaned_df.dropna(
    subset=["fare", "distance", "pickup_borough"]
)
````

### What this output means

Rows are removed only when `fare`, `distance`, or `pickup_borough` is missing.

Missing values in unrelated columns do not cause the row to be removed.

## 41.4 Check the important columns again

### Before you run the code

Now we'll verify that those three columns no longer contain missing values.

````python
cleaned_df[
    ["fare", "distance", "pickup_borough"]
].isna().sum()
````

### What this output means

You should see `0` for all three columns.

That confirms that the rows with missing values in these important fields were removed successfully.

## 41.5 Reset the index

### Before you run the code

After removing rows, the DataFrame index can contain gaps.

For example, it might look like:

```text
0
1
4
6
7
```

`reset_index()` creates a fresh continuous index.

`drop=True` means pandas should not keep the old index as another column.

````python
cleaned_df = cleaned_df.reset_index(drop=True)
````

### What this output means

The cleaned DataFrame now has a fresh row index:

```text
0, 1, 2, 3, 4, ...
```

Only the row labels changed. The taxi-trip data itself did not.

## 41.6 Preview the cleaned data

### Before you run the code

Finally, let's inspect the first few rows to make sure the cleaned dataset still looks correct.

````python
cleaned_df.head()
````

### What this output means

This shows the first five rows of the cleaned DataFrame.

At this point, our cleaning flow was:

```text
copy original data
      ↓
remove duplicates
      ↓
remove rows missing important values
      ↓
verify missing values
      ↓
reset row index
      ↓
inspect the cleaned result
```

# 42. Save cleaned data

### Before you run the code

`to_csv()` writes a DataFrame to a CSV file.

````python
cleaned_df.to_csv("cleaned_taxis.csv", index=False)
````

### What this output means

`cleaned_taxis.csv` is now saved in the Colab environment. `index=False` avoids writing the pandas index as an extra column.
