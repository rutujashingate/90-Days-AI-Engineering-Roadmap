# 2. Mean and Median - What Is a Typical Value?

Suppose we want to know:

> **What was a typical Titanic ticket fare?**

### Mean

**Definition:** The **mean** is the arithmetic average.

```text
10, 20, 30, 40, 50

mean = 30
```

### Median

**Definition:** The **median** is the middle value after sorting.

```text
10, 20, 30, 40, 500

median = 30
```

The extreme value `500` pulls the mean upward, but the median stays near the middle.

### Why useful in AI?

Use mean/median to summarize:

```text
feature values
model latency
agent cost
average loss
user/session metrics
```

Median is especially useful when a few extreme values would distort the mean.

### In AI

```text
Mean / median
-> typical latency, cost, model score, or retrieval score
```

Use these when you want to know what a **typical value** looks like.

````python
# Keep only known fare values.
fares = df["fare"].dropna()

# Average fare.
mean_fare = fares.mean()

# Middle fare after sorting.
median_fare = fares.median()

print("Mean fare  :", mean_fare)
print("Median fare:", median_fare)
````

````python
# Create a new graph that is 8 inches wide and 5 inches tall.
plt.figure(figsize=(8, 5))

# A histogram groups fare values into ranges called bins.
# hist() groups fares into ranges and counts how many fall in each range.
# bins=30 means use 30 fare ranges.
plt.hist(fares, bins=30, edgecolor="black", alpha=0.75)

# Show where the mean is.
plt.axvline(
    mean_fare,
    linestyle="--",
    linewidth=2,
    label=f"Mean = {mean_fare:.2f}"
)

# Show where the median is.
plt.axvline(
    median_fare,
    linestyle=":",
    linewidth=2,
    label=f"Median = {median_fare:.2f}"
)

plt.xlabel("Ticket fare")
plt.ylabel("Number of passengers")
plt.title("Titanic Ticket Fare Distribution")
plt.legend()
plt.grid(axis="y", alpha=0.3)
plt.show()
````

### What this graph shows

- **x-axis:** ticket fare
- **y-axis:** number of passengers in each fare range
- **bars:** how common each fare range was
- **mean line:** average fare
- **median line:** middle fare

The distribution has a long right tail because a small number of passengers paid very high fares. That is why the mean can be noticeably higher than the median.

### AI use

Mean and median are useful for:

- understanding numerical features
- filling missing values
- detecting unusual values
- feature engineering
- monitoring model metrics

# 3. Variance and Standard Deviation - How Spread Out Is the Data?

The average alone does not tell us how consistent the values are.

```text
A = 29, 30, 30, 31
B = 5, 20, 40, 55
```

Both can have a similar center, but B is much more spread out.

### Variance

**Definition:** Variance measures spread around the mean using squared distances.

$$
\mathrm{Var}(X)=\frac{1}{N}\sum_{i=1}^{N}(x_i-\mu)^2
$$

### Standard deviation

**Definition:** Standard deviation is the square root of variance.

It is easier to read because it uses the same units as the original feature.

```text
small std -> values are fairly consistent
large std -> values vary a lot
```

### Why useful in AI?

It helps detect:

```text
noisy features
unstable latency
unstable model scores
outliers
data drift
```

### In AI

```text
Variance / standard deviation
-> how stable or inconsistent those values are
```

Use these when the average alone is not enough and you need to know **how much the values vary**.

````python
# Keep only known ages.
ages = df["age"].dropna()

# ddof=0 calculates population-style variance/std for these values.
age_variance = ages.var(ddof=0)
age_std = ages.std(ddof=0)

print("Mean age:", ages.mean())
print("Variance:", age_variance)
print("Standard deviation:", age_std)
````

````python
mean_age = ages.mean()
std_age = ages.std(ddof=0)

plt.figure(figsize=(8, 5))
plt.hist(ages, bins=30, edgecolor="black", alpha=0.75)

# Mean age.
plt.axvline(mean_age, linewidth=2, label="Mean")

# One standard deviation below and above the mean.
plt.axvline(mean_age - std_age, linestyle="--", label="Mean - 1 std")
plt.axvline(mean_age + std_age, linestyle="--", label="Mean + 1 std")

plt.xlabel("Age")
plt.ylabel("Number of passengers")
plt.title("Passenger Ages and Standard Deviation")
plt.legend()
plt.grid(axis="y", alpha=0.3)
plt.show()
````

### What this graph shows

- **bars:** passenger age distribution
- **middle line:** mean age
- **outer dashed lines:** one standard deviation below and above the mean

Standard deviation gives us one number describing the typical scale of spread around the mean.

### AI use

Variance and standard deviation help with:

- feature scaling
- outlier detection
- data-drift monitoring
- comparing feature variability

# 4. Probability - How Likely Is an Event?

**Definition:** Probability measures how likely an event is.

```text
0   -> impossible
0.5 -> 50% chance
1   -> certain
```

### Titanic example

```text
P(Survived)
=
number of survivors
/
total passengers
```

### Why useful in AI?

Many AI systems output probabilities:

```text
P(spam) = 0.92
P(fraud) = 0.08
P(class = dog) = 0.74
```

Probability lets us represent uncertainty instead of pretending every prediction is certain.

### In AI

```text
Probability
-> chance of success, failure, fraud, spam, or a class label
```

Use probability when the outcome is uncertain and you want to measure **how likely** it is.

````python
# survived is 1 for survivors and 0 otherwise.
survivors = df["survived"].sum()
total_passengers = len(df)

p_survived = survivors / total_passengers

print("Survivors:", survivors)
print("Total passengers:", total_passengers)
print("P(Survived):", p_survived)
````

Because `survived` contains only `0` and `1`, its mean gives the same probability.

````python
print("Mean of survived column:", df["survived"].mean())
````

Why?

For binary values such as:

```text
0, 1, 0, 1, 1
```

the mean is simply:

```text
number of 1s
÷
total values
```

So the mean of a binary indicator equals the observed probability of `1`.

# 7. Conditional Probability - Probability After We Know Something

**Definition:** Conditional probability is the probability of an event after we already know another fact.

$$
P(A \mid B)
$$

The `|` means:

> **given that**

Titanic example:

```text
P(Survived)
```

asks:

> What fraction of all passengers survived?

But:

```text
P(Survived | Female)
```

asks:

> Among female passengers, what fraction survived?

### Why useful in AI?

Predictions are usually conditional on available information:

```text
P(Fraud | transaction features)

P(Spam | email content)

P(User clicks | recommendation shown)
```

### In AI

```text
Conditional probability
-> success given task type, tool, user group, or other evidence
```

Use this when the probability changes after you know something extra.

````python
# Filter to female passengers.
female_passengers = df[df["sex"] == "female"]

# Filter to male passengers.
male_passengers = df[df["sex"] == "male"]

# Conditional probabilities.
p_survived_given_female = female_passengers["survived"].mean()
p_survived_given_male = male_passengers["survived"].mean()

print("P(Survived | Female):", p_survived_given_female)
print("P(Survived | Male)  :", p_survived_given_male)
````

````python
groups = ["All", "Female", "Male"]

probabilities = [
    df["survived"].mean(),
    p_survived_given_female,
    p_survived_given_male
]

plt.figure(figsize=(8, 5))
plt.bar(groups, probabilities)
plt.ylim(0, 1)

plt.ylabel("Survival probability")
plt.title("Overall vs Conditional Probability")
plt.grid(axis="y", alpha=0.3)
plt.show()
````

### What this graph shows

The probability changes when we condition on additional information.

```text
P(Survived)
```

and:

```text
P(Survived | Female)
```

are different questions.

# 8. Bayes' Theorem - Update Probability With New Evidence

**Definition:** Bayes' theorem is a mathematical rule for **updating the probability of an event when new evidence becomes available**.

The easiest way to think about it is:

Bayes is easiest to think about as:

```text
starting probability
+
new evidence
↓
updated probability
```

$$
P(A \mid B)
=
\frac{P(B \mid A)P(A)}{P(B)}
$$

Titanic example:

```text
A = Survived
B = Female
```

We want:

```text
P(Survived | Female)
```

### Prior

Probability **before** using the new evidence.

```text
P(Survived)
```

### Likelihood

How common the evidence is if the hypothesis is true.

```text
P(Female | Survived)
```

### Posterior

Updated probability **after** using the evidence.

```text
P(Survived | Female)
```

### Why useful in AI?

Bayesian reasoning is useful whenever new evidence should change uncertainty:

```text
diagnosis
spam detection
probabilistic models
decision systems
uncertainty estimation
```

### In AI

```text
Bayes
-> update probability after new evidence
```

Use Bayes when you already have a probability and want to **update it after seeing new information**.

````python
# PRIOR:
# Probability of survival before using the evidence "female".
prior = df["survived"].mean()

# LIKELIHOOD:
# Among passengers who survived, how common is the evidence "female"?
survived_passengers = df[df["survived"] == 1]
likelihood = (survived_passengers["sex"] == "female").mean()

# Probability of observing the evidence itself.
p_female = (df["sex"] == "female").mean()

# BAYES' THEOREM.
posterior = (likelihood * prior) / p_female

print("Prior P(Survived):", prior)
print("Likelihood P(Female | Survived):", likelihood)
print("P(Female):", p_female)
print("Posterior P(Survived | Female):", posterior)
````

### Verify it directly

If Bayes' theorem is used correctly, the result should match the conditional probability obtained by filtering the dataset directly.

````python
direct_probability = female_passengers["survived"].mean()

print("Bayes result :", posterior)
print("Direct result:", direct_probability)
````

# 9. Expectation - What Average Outcome Should We Expect?

**Definition:** Expected value is the probability-weighted average outcome over many repetitions.

For:

```text
S = 1 if survived
S = 0 if not
```

$$
E[S]
=
1 \times P(S=1)
+
0 \times P(S=0)
$$

So:

$$
E[S]=P(Survived)
$$

An expected value of `0.38` does **not** mean one passenger was "38% survived."

It means the average of many `0/1` outcomes approaches `0.38`.

### Why useful in AI?

Expectation appears in:

```text
expected loss
expected reward
risk
reinforcement learning
optimization
```

Many training objectives are averages across examples.

### In AI

```text
Expectation
-> expected loss, reward, or cost
```

Use expectation when you want the **average outcome you should expect over many runs**.

````python
expected_survival = df["survived"].mean()
expected_fare_estimate = df["fare"].mean()

print("Expected survival variable:", expected_survival)
print("Estimated expected fare:", expected_fare_estimate)
````

```text
Population vs sample

```

# 12. Random Sampling - Different Samples Give Different Answers

**Definition:** Random sampling selects observations using a random process.

It reduces the effect of deliberately choosing only convenient examples.

But different random samples can still produce slightly different statistics.

That normal difference is called **sampling variation**.

### AI example

Two random evaluation sets might give:

```text
Agent success rate on sample A = 84%
Agent success rate on sample B = 89%
```

Some difference is expected because the samples are different.

### In AI

```text
Random sampling
-> choose evaluation examples more fairly
```

Use random sampling so you do not accidentally select only examples that make the system look good.

````python
# Reproducible random generator.
rng = np.random.default_rng(42)

# Convert ages to a NumPy array.
age_values = ages.to_numpy()

sample_means = []

# Take 300 random samples of 80 passengers each.
for _ in range(300):
    sample = rng.choice(
        age_values,
        size=80,
        replace=False
    )

    sample_means.append(sample.mean())

sample_means = np.array(sample_means)
````

````python
plt.figure(figsize=(8, 5))

# Each value here is the mean from one random sample.
# The histogram shows how those sample means vary.
plt.hist(
    sample_means,
    bins=25,
    edgecolor="black",
    alpha=0.75
)

plt.axvline(
    population_mean_age,
    linestyle="--",
    linewidth=2,
    label="Full-dataset mean"
)

plt.xlabel("Mean age from one random sample")
plt.ylabel("Number of samples")
plt.title("Random Samples Produce Slightly Different Means")
plt.legend()
plt.grid(axis="y", alpha=0.3)
plt.show()
````

### What this graph shows

Each value in the histogram is the mean from a different random sample.

The answers vary slightly, but many cluster around the full-dataset mean.

That is **sampling variability**.

# 13. Sampling Bias - A Bad Sample Can Mislead You

**Definition:** Sampling bias happens when the selection method systematically over-represents or under-represents part of the population.

Titanic example:

```text
measure overall survival
but sample only first-class passengers
```

The calculation may be correct for first class, but the sample does not represent everyone.

### AI example

Suppose an agent evaluation contains:

```text
90 easy tasks
10 hard multi-step tasks
```

A high overall score may hide poor performance on the hard tasks.

### Why useful in AI?

Sampling bias can create:

```text
misleading model metrics
misleading agent evals
poor generalization
systematic failures
bad product decisions
```

### In AI

```text
Sampling bias
-> catch misleading evals caused by an unrepresentative sample
```

Use this idea when an evaluation score looks good but the test data may not represent real usage.

````python
full_survival_rate = df["survived"].mean()
full_mean_fare = df["fare"].mean()

# Intentionally biased sample.
biased_sample = df[df["pclass"] == 1]

biased_survival_rate = biased_sample["survived"].mean()
biased_mean_fare = biased_sample["fare"].mean()

print("Full survival rate      :", full_survival_rate)
print("First-class-only rate   :", biased_survival_rate)

print()
print("Full mean fare          :", full_mean_fare)
print("First-class-only mean fare:", biased_mean_fare)
````

````python
labels = [
    "Full dataset",
    "First-class-only sample"
]

values = [
    full_survival_rate,
    biased_survival_rate
]

plt.figure(figsize=(8, 5))
plt.bar(labels, values)
plt.ylim(0, 1)

plt.ylabel("Survival rate")
plt.title("Sampling Bias Can Change the Conclusion")
plt.grid(axis="y", alpha=0.3)
plt.show()
````

### Why this matters in AI

Biased training or evaluation data can cause:

- misleading metrics
- poor generalization
- systematic errors
- bad product decisions

A strong model cannot automatically fix a badly biased dataset.

# 14. Bootstrap Intuition - How Stable Is Our Estimate?

Suppose we report:

```text
Agent success rate = 87%
```

One number does not tell us how much that estimate might change with a different evaluation sample.

**Definition:** Bootstrap repeatedly resamples the observed data **with replacement** and recalculates the statistic.

With replacement means the same observation can be selected more than once.

Example:

```text
original:
A B C D

bootstrap sample:
A C C D
```

Process:

```text
observed data
↓
resample with replacement
↓
calculate statistic
↓
repeat many times
↓
inspect how much the statistic varies
```

### Why useful in AI?

Bootstrap can help study uncertainty around:

```text
accuracy
agent success rate
precision / recall
latency
cost
other evaluation metrics
```

### In AI

```text
Bootstrap
-> estimate how much a metric could change with another sample
```

Use bootstrap when you want to know whether a metric such as `87% success` is stable or could move noticeably with a different evaluation set.

````python
# Reproducible random generator.
rng = np.random.default_rng(42)

age_values = ages.to_numpy()
bootstrap_means = []

# Create 1000 bootstrap samples.
for _ in range(1000):
    bootstrap_sample = rng.choice(
        age_values,
        size=len(age_values),
        replace=True
    )

    # Save the mean from this bootstrap sample.
    bootstrap_means.append(
        bootstrap_sample.mean()
    )

bootstrap_means = np.array(bootstrap_means)

# Approximate central 95% percentile interval.
lower, upper = np.percentile(
    bootstrap_means,
    [2.5, 97.5]
)

print("Observed mean age:", age_values.mean())
print("Approx. 95% bootstrap interval:", lower, "to", upper)
````

````python
plt.figure(figsize=(8, 5))

# Each value is the mean from one bootstrap resample.
# The histogram shows how much the estimate changes across resamples.
plt.hist(
    bootstrap_means,
    bins=30,
    edgecolor="black",
    alpha=0.75
)

plt.axvline(
    age_values.mean(),
    linewidth=2,
    label="Observed mean"
)
plt.axvline(
    lower,
    linestyle="--",
    label="2.5th percentile"
)
plt.axvline(
    upper,
    linestyle="--",
    label="97.5th percentile"
)

plt.xlabel("Bootstrap mean age")
plt.ylabel("Number of bootstrap samples")
plt.title("Bootstrap Distribution of Mean Passenger Age")
plt.legend()
plt.grid(axis="y", alpha=0.3)
plt.show()
````

### What this graph shows

Each bootstrap sample gives a slightly different mean.

The spread of bootstrap means gives intuition for the uncertainty in our estimate.

For Day 8, remember:

> **Bootstrap = repeatedly resample with replacement to study how a statistic might vary.**

```text
api             0             6.5
api             1             4.0
multi_step      0             9.2
multi_step      1             7.8
```

From this small table we can ask:

```text
What is the overall success probability?
What is P(success | task type)?
What is typical latency?
How variable is latency?
```

````python
# Small illustrative agent-evaluation table.
agent_eval = pd.DataFrame({
    "task_type": [
        "search",
        "search",
        "api",
        "api",
        "multi_step",
        "multi_step"
    ],
    "success": [1, 1, 0, 1, 0, 1],
    "latency_seconds": [2.1, 1.8, 6.5, 4.0, 9.2, 7.8]
})

agent_eval
````

````python
# Overall probability of success.
overall_success = agent_eval["success"].mean()

# Conditional success probability by task type.
success_by_task = (
    agent_eval
    .groupby("task_type")["success"]
    .mean()
)

# Center and spread of latency.
mean_latency = agent_eval["latency_seconds"].mean()
median_latency = agent_eval["latency_seconds"].median()
std_latency = agent_eval["latency_seconds"].std(ddof=0)

print("Overall success:", overall_success)
print()
print("Success by task type:")
print(success_by_task)
print()
print("Mean latency:", mean_latency)
print("Median latency:", median_latency)
print("Latency standard deviation:", std_latency)
````

### What this shows

The same statistics answer practical engineering questions:

```text
mean / median
-> what is typical?

standard deviation
-> how consistent is the system?

probability
-> how often does it succeed?

conditional probability
-> which task types are weak?

sampling
-> does the eval represent real usage?

bootstrap
-> how stable is the reported metric?
```

That is the connection between statistics and building AI systems.
