# 2. Dot Product - Turn Two Vectors Into One Score

**Definition:** A **dot product** takes two equal-length vectors, multiplies their matching values, adds the results, and produces **one number**.

A useful mental model:

> **two vectors → one score**

That score is later used in similarity, attention, recommendations, and matrix multiplication.


## The problem

Suppose we have:

```text
A = [1, 2]
B = [3, 4]
```

We want one number that summarizes how these two vectors interact.

The dot product does this:

```text
1×3 = 3
2×4 = 8

3 + 8 = 11
```

So the dot product is `11`.

Plain-English rule:

> Multiply matching positions, then add the results.

````python
a = np.array([1, 2])
b = np.array([3, 4])

print("Dot product:", np.dot(a, b))
````

### Why this matters

The dot product later appears in:

- similarity scoring
- attention
- recommendation systems
- neural networks
- matrix multiplication

# 3. Vector Norm - How Long Is the Vector?

**Definition:** A **vector norm** measures the **size or length (magnitude)** of a vector.

A useful mental model:

> **norm = how long is this vector?**

For example, `[3, 4]` has length `5`.


## The problem

Take:

```text
v = [3, 4]
```

Imagine it as an arrow from `(0,0)` to `(3,4)`.

Its length is:

```text
√(3² + 4²)
= √25
= 5
```

That length is called the **norm**.

````python
v = np.array([3, 4])

print("Vector:", v)
# np.linalg.norm(v) uses NumPy's linear algebra tools to calculate the length of vector v
print("Norm:", np.linalg.norm(v))
````

### Why this matters

Vectors can point in the same direction but have very different sizes:

```text
[1, 1]
[10, 10]
```

Norms help us separate **size** from **direction**.

That becomes important for cosine similarity.

# 4. Cosine Similarity - Are Two Vectors Pointing in the Same Direction?

**Definition:** **Cosine similarity measures how similarly two vectors point, regardless of how large the vectors are.**

The score is mainly used to rank and choose the most relevant items.

A useful mental model:

> **same direction → high similarity**  
> **different direction → lower similarity**

In AI, an embedding model may represent text, images, users, or products as vectors. Cosine similarity then compares those vectors.

Important: cosine similarity itself does **not** understand meaning. The embedding model creates the numerical representation; cosine similarity only compares the resulting vector directions.


This is one of the most useful ideas for embeddings, RAG, and semantic search.

## The problem

Suppose an embedding model gives us:

```text
"Learn Python"      → [0.9, 1.1]
"Python tutorial"   → [1.0, 1.0]
"Cooking pasta"     → [1.0, -1.0]
```

We want to answer:

> Which document is most similar to `"Learn Python"`?

The idea:

> If two vectors point in a similar direction, treat them as more similar.

````python
vectors = {
    "Learn Python": np.array([0.9, 1.1]),
    "Python tutorial": np.array([1.0, 1.0]),
    "Cooking pasta": np.array([1.0, -1.0]),
}

plt.figure(figsize=(6, 6))
plt.axhline(0)
plt.axvline(0)

for name, v in vectors.items():
    plt.arrow(
        0, 0, v[0], v[1],
        head_width=0.06,
        length_includes_head=True
    )
    plt.text(v[0] * 1.08, v[1] * 1.08, name)

plt.xlim(-1.5, 1.5)
plt.ylim(-1.5, 1.5)
plt.grid(True)
plt.title("Direction matters")
plt.show()
````

### What should you notice?

`"Learn Python"` and `"Python tutorial"` point almost the same way.

`"Cooking pasta"` points in a very different direction.

Cosine similarity gives us a number for this.

Typical meaning:

```text
close to 1   → very similar direction
close to 0   → not strongly related
close to -1  → opposite direction
```

The formula is:

\[
\text{cosine similarity}
=
\frac{a\cdot b}{\|a\|\|b\|}
\]

Do not memorize it yet.

Think of it as:

```text
dot product
÷
both vector lengths
```

The division removes the effect of vector size and focuses on direction.

````python
query = np.array([0.9, 1.1])
python_doc = np.array([1.0, 1.0])

dot = np.dot(query, python_doc)
query_length = np.linalg.norm(query)
doc_length = np.linalg.norm(python_doc)

similarity = dot / (query_length * doc_length)

print("Dot product:", dot)
print("Query length:", query_length)
print("Document length:", doc_length)
print("Cosine similarity:", similarity)
````

### What the output means

The score is very close to `1`.

So these two vectors point in nearly the same direction.

In an embedding system, that suggests the texts are semantically similar.

````python
def cosine_similarity(a, b):
    return np.dot(a, b) / (
        np.linalg.norm(a) * np.linalg.norm(b)
    )

query = np.array([1.0, 0.9])

documents = {
    "Python basics": np.array([1.0, 1.0]),
    "Machine learning": np.array([0.8, 1.0]),
    "Travel guide": np.array([-0.8, 0.1]),
}

for name, vector in documents.items():
    print(name, "→", round(cosine_similarity(query, vector), 3))
````

### Where you'll use this

- RAG retrieval
- semantic search
- similar-document search
- recommendation systems
- embedding comparisons

# 5. Linear Transformation - A Matrix as a Machine

**Definition:** A **linear transformation** is a rule represented by a matrix that changes an input vector into a new vector while preserving linear relationships.

A useful mental model:

> **vector in → matrix transformation → new vector out**

The transformation can stretch, shrink, rotate, reflect, or mix information.


## The problem

Suppose we have a point:

```text
[2, 1]
```

and we want to:

```text
double x
leave y unchanged
```

A matrix can store that rule.

Think:

```text
input vector
     ↓
transformation matrix
     ↓
output vector
```

````python
point = np.array([2, 1])

transform = np.array([
    [2, 0],
    [0, 1]
])

new_point = transform @ point

print("Before:", point)
print("After :", new_point)
````

### What happened?

```text
[2, 1] → [4, 1]
```

The first coordinate doubled.

The second stayed the same.

That is a linear transformation.

````python
plt.figure(figsize=(6, 6))
plt.axhline(0)
plt.axvline(0)

plt.arrow(0, 0, point[0], point[1],
          head_width=0.12, length_includes_head=True)

plt.arrow(0, 0, new_point[0], new_point[1],
          head_width=0.12, length_includes_head=True)

plt.xlim(-1, 5)
plt.ylim(-1, 3)
plt.grid(True)
plt.title("Before and after transformation")
plt.show()
````

### Where you'll use this

Neural networks repeatedly do:

```text
input
  ↓
weight matrix
  ↓
new representation
```

Transformers also use matrix transformations for queries, keys, values, and projections.

# 6. Linear Systems - \(Ax=b\)

**Definition:** A **linear system** is a collection of linear equations that must all be true at the same time.

`Ax = b` is a compact way to write that system:

- `A` = known coefficients
- `x` = unknown values we want to find
- `b` = known results

A useful mental model:

> **find the unknown input `x` that makes `A @ x` produce `b`**


The notation looks scary, so start with a normal problem.

## Cafe example

```text
2 coffees + 1 sandwich = $11
1 coffee  + 2 sandwiches = $13
```

Let:

```text
x = coffee price
y = sandwich price
```

Then:

\[
2x+y=11
\]

\[
x+2y=13
\]

We can store this as:

```text
A = coefficients
x = unknown prices
b = totals
```

So:

\[
Ax=b
\]

means:

> Find the unknown vector `x` that makes `A @ x` equal `b`.

````python
A = np.array([
    [2, 1],
    [1, 2]
])

b = np.array([11, 13])

solution = np.linalg.solve(A, b)

print("Coffee:", solution[0])
print("Sandwich:", solution[1])
````

Verify the result:

````python
print("A @ solution:", A @ solution)
print("Expected b  :", b)
````

### Where you'll use this

The idea appears underneath:

- regression
- optimization
- scientific computing
- computer vision
- numerical methods

# 7. Rank - How Much Independent Information Is Really Here?

**Definition:** The **rank of a matrix** tells us how much **unique or independent information** the matrix contains.

An **independent direction** means a row or column that cannot be recreated just by multiplying or combining the others.

If one row or column can be created from another, it is **redundant** and does not add new information.

> **Rank = how many pieces of information are actually unique, rather than repeated or dependent on each other.**

A useful mental model:

> **rank = how much genuinely new information is here?**


## The problem

Look at:

```text
Row 1 = [1, 2]
Row 2 = [2, 4]
```

At first, there are two rows.

But:

```text
[2, 4] = 2 × [1, 2]
```

So Row 2 gives us no new direction.

It is redundant.

Rank tells us roughly:

> How many independent directions or pieces of information are in this matrix?

````python
M = np.array([
    [1, 2],
    [2, 4]
])

print("Rank:", np.linalg.matrix_rank(M))
````

Even though there are two rows, the rank is `1`.

Now compare:

````python
M2 = np.array([
    [1, 2],
    [3, 5]
])

print("Rank:", np.linalg.matrix_rank(M2))
````

This one has rank `2` because the second row is not just a scaled copy.

### Why rank matters

- detecting redundant features
- dimensionality reduction
- compression
- recommendation systems
- understanding LoRA
- knowing whether information was lost

# 8. Invertibility - Can We Undo a Transformation?

**Definition:** A matrix is **invertible** when its transformation can be **uniquely reversed**.

Its inverse acts like an undo operation:

> **transform → inverse transform → original value**

If the original transformation loses information, there may be no unique way to reverse it.


Suppose:

```text
input
  ↓
matrix
  ↓
output
```

If we can uniquely reverse that process, the matrix is **invertible**.

The inverse is written:

\[
A^{-1}
\]

and it acts like an undo operation.

````python
A = np.array([
    [2.0, 1.0],
    [1.0, 1.0]
])

A_inv = np.linalg.inv(A)

print("A:")
print(A)

print("\nInverse:")
print(A_inv)

print("\nA @ inverse:")
print(A @ A_inv)
````

The result is approximately:

```text
[[1, 0],
 [0, 1]]
```

That is the identity matrix.

It means the inverse successfully cancels the original transformation.

Now try a matrix with redundant information:

````python
bad = np.array([
    [1.0, 2.0],
    [2.0, 4.0]
])

print("Rank:", np.linalg.matrix_rank(bad))

try:
    print(np.linalg.inv(bad))
except np.linalg.LinAlgError:
    print("Cannot invert this matrix.")
````

Why?

Because information was collapsed.

There is no unique way to go backward.

Simple intuition:

```text
full independent information
→ may be reversible

lost / duplicated information
→ cannot uniquely reverse
```

# 9. SVD - Find the Strongest Hidden Patterns

**Definition:** **SVD (Singular Value Decomposition)** breaks a matrix into simpler pieces that reveal its most important underlying directions or patterns.

The singular values tell us how strong those patterns are:

```text
large singular value → strong pattern
small singular value → weaker pattern
```

A useful mental model:

> **SVD = break a complicated matrix into important patterns and their strengths**


**SVD = Singular Value Decomposition**

Forget the formula for a moment.

## The problem

Imagine a movie app has this table:

```text
            Action A  Action B  Romance C
User 1          5         4         1
User 2          4         5         1
User 3          1         1         5
User 4          1         1         4
```

You can see a hidden pattern:

```text
Users 1 and 2 → prefer action
Users 3 and 4 → prefer romance
```

SVD helps a computer find strong hidden patterns like these inside a matrix.

````python
ratings = np.array([
    [5.0, 4.0, 1.0],
    [4.0, 5.0, 1.0],
    [1.0, 1.0, 5.0],
    [1.0, 1.0, 4.0]
])

print(ratings)
````

SVD breaks a matrix into:

\[
A=U\Sigma V^T
\]

You do not need to memorize those pieces yet.

Start with the singular values:

```text
large value  → strong pattern
small value  → weaker pattern
```

````python
U, S, Vt = np.linalg.svd(
    ratings,
    full_matrices=False
)

print("Singular values:")
print(S)
````

Now rebuild the original matrix:

````python
reconstructed = U @ np.diag(S) @ Vt

print("Original:")
print(ratings)

print("\nReconstructed:")
print(reconstructed)
````

The reconstruction is almost identical.

So SVD did not destroy the information.

It reorganized the matrix so its important directions are easier to work with.

### Where you'll use SVD

- recommendation systems
- PCA
- compression
- noise reduction
- latent-pattern discovery
- low-rank approximation

# 10. Low-Rank Approximation - Keep Only the Strongest Patterns

**Definition:** A **low-rank approximation** creates a simpler version of a matrix by keeping its strongest patterns and discarding weaker ones.

A useful mental model:

> **keep the important structure, remove less important detail**

The result is smaller/simpler but only an approximation of the original matrix.


Now suppose SVD finds:

```text
Pattern 1 → strong
Pattern 2 → useful
Pattern 3 → weak
```

Maybe we don't need all three.

We can keep only the strongest patterns.

That gives a **low-rank approximation**.

Think:

```text
full matrix
   ↓
keep strongest structure
   ↓
smaller approximation
```

````python
U, S, Vt = np.linalg.svd(
    ratings,
    full_matrices=False
)

k = 1

rank1 = (
    U[:, :k]
    @ np.diag(S[:k])
    @ Vt[:k, :]
)

print("Original:")
print(ratings)

print("\nRank-1 approximation:")
print(rank1)
````

The approximation is not exact because we deliberately threw away weaker patterns.

Now keep two:

````python
k = 2

rank2 = (
    U[:, :k]
    @ np.diag(S[:k])
    @ Vt[:k, :]
)

print(rank2)
````

Simple tradeoff:

```text
smaller k
→ simpler, but less exact

larger k
→ keeps more information
```

### Where you'll use this

- recommendation systems
- compression
- denoising
- model compression
- LoRA intuition

# 11. PCA - Can We Describe the Same Data With Fewer Numbers?

**Definition:** **PCA (Principal Component Analysis)** reduces the number of features/dimensions in data by finding new directions that capture the most important variation.

A useful mental model:

> **many related features → fewer new features that preserve the main structure**

For example, if two features mostly tell the same story, PCA may summarize them using one principal direction.


**PCA = Principal Component Analysis**

## The problem

Suppose we track:

```text
hours studied
practice questions completed
```

Example:

```text
1 hour  → 10 questions
2 hours → 19 questions
3 hours → 31 questions
4 hours → 39 questions
5 hours → 52 questions
```

These two features are strongly related.

So perhaps most of the information follows one main direction.

PCA tries to find that main direction.

````python
X = np.array([
    [1.0, 10.0],
    [2.0, 19.0],
    [3.0, 31.0],
    [4.0, 39.0],
    [5.0, 52.0]
])

plt.figure(figsize=(6, 5))
plt.scatter(X[:, 0], X[:, 1])
plt.xlabel("Hours studied")
plt.ylabel("Practice questions")
plt.grid(True)
plt.title("Two related features")
plt.show()
````

The points mostly follow one direction.

PCA asks:

> Can we represent this pattern using one number instead of two?

## Step 1 — Center the data

Subtract the mean of each feature.

````python
mean = X.mean(axis=0)
X_centered = X - mean

print("Mean:", mean)
print("\nCentered data:")
print(X_centered)
````

## Step 2 — Find the strongest direction with SVD

````python
U, S, Vt = np.linalg.svd(
    X_centered,
    full_matrices=False
)

print("Singular values:", S)
print("Directions:")
print(Vt)
````

## Step 3 — Keep only the strongest direction

````python
first_direction = Vt[0]

X_1d = X_centered @ first_direction

print("Original shape:", X.shape)
print("Reduced shape :", X_1d.shape)
print("\nReduced data:")
print(X_1d)
````

### What happened?

Before:

```text
one student → [hours, questions]
```

Two numbers.

After PCA:

```text
one student → [principal value]
```

One number.

We reduced:

```text
2 dimensions → 1 dimension
```

while keeping the strongest pattern in the data.

### Where you'll use PCA

- reducing many numerical features
- visualization
- preprocessing
- removing correlated information
- exploratory analysis
- anomaly detection
